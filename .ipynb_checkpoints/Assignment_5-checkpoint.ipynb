#### Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like: 

```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
```

##### a) Addition of 2 in all the element b) Multiply 3 with all the elements c) Divide every element by 2

```python
print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

#### Q.2 Questions on Basic NumPy Array: 
##### a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5]) 
##### b) Find the most frequent value and their indice(s) in the following arrays: 
###### i. x = np.array([1,2,3,4,5,1,2,1,1,1]) 
###### ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
arr = np.array([1,2,3,6,4,5])
print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

#### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute? 

```python
arr = np.linspace(10,100,25)
print(arr)
```

```python
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Total elements:", arr.size)
print("Data type:", arr.dtype)
print("Total bytes:", arr.nbytes)
```

```python
transposed_arr = arr.reshape(5, 5)

print("Reshaped array:")
print(transposed_arr)
```

```python
transposed_arr = arr.reshape(5, 5)

print("Transpose using .T:")
print(transposed_arr.T)
```

#### Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_samriddhi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print(ucs420_samriddhi)
```

```python
print("Mean:", np.mean(ucs420_samriddhi))
print("Median:", np.median(ucs420_samriddhi))
print("Maximum:", np.max(ucs420_samriddhi))
print("Minimum:", np.min(ucs420_samriddhi))
print("Unique elements:", np.unique(ucs420_samriddhi))
```

```python
reshaped_ucs420_samriddhi = ucs420_samriddhi.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_samriddhi)
```

```python
resized_ucs420_samriddhi = np.resize(ucs420_samriddhi, (2, 3))

print("Resized array:")
print(resized_ucs420_samriddhi)
```
