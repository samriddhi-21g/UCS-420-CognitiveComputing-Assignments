#### Q1. Ask the user to enter a value (e.g., M) for a mathematical function. 
#####  Generate x values from -10 to 10 using np.linspace(). 
#####  Compute y values for: 
###### o y= M⋅x2 
###### o y=M⋅sin(x) 
#####  Plot both functions in a single figure: 
###### o Use different colors and line styles. 
###### o Add legend, grid, and title. 

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
M = float(input("Enter the value of M: "))
x = np.linspace(-10,10,100)
print("M =", M)
print("x values:", x)
```

```python
y1 = M*x**2
y2 = M*np.sin(x)

print(y1)
print(y2)
```

```python
plt.plot(x,y1,color="blue",linestyle="-",label="y = Mx2")
plt.plot(x,y2,color="red",linestyle="--",label="y = M sin(x)")

plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid()

plt.show()
```

#### Q.2 Create a dataset of five subjects and scores. 
#####  Convert it into a Pandas DataFrame. 
#####  Plot the scores using a Seaborn bar plot with: 
###### o Different colors for each bar. 
###### o Annotations on top of each bar. 
###### o Title, axis labels, and grid.

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
data = {
    "Subject": ["Python", "Java", "DBMS", "AI", "Mathematics"],
    "Score": [85, 78, 90, 88, 82]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x="Subject", y="Score", data=df, hue="Subject", legend=False)

for i, score in enumerate(df["Score"]):
    plt.text(i, score + 1, str(score), ha="center")

plt.title("Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")

plt.show()
```

#### Q3. Write a Python script to select your roll number as seed for NumPy and then generate a dataset of 50 values using np.random.randn(). 
#####  Create a 2x2 subplot layout displaying: 
###### o Line plot showing cumulative sum. 
###### o Scatter plot with random noise. 
#####  Customize with titles, axis labels, and grids. 

```python
np.random.seed(65)
data = np.random.randn(50)
print(data)
```

```python
# Plot -1 
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)

cumulative_sum = np.cumsum(data)

plt.plot(cumulative_sum)

plt.title("Cumulative Sum")
plt.xlabel("Index")
plt.ylabel("Cumulative Sum")
plt.grid()

# Plot - 2
noise = np.random.randn(50)

plt.subplot(2, 2, 2)

plt.scatter(range(50), data + noise)

plt.title("Scatter Plot with Random Noise")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid()

# Plot 3 - Original Data
plt.subplot(2, 2, 3)

plt.plot(data)

plt.title("Original Random Data")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid()


# Plot 4 - Histogram
plt.subplot(2, 2, 4)

plt.hist(data)

plt.title("Distribution of Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid()


plt.tight_layout()
plt.show()
```

#### Q.4 Download Data-set from the below link 
###### https://github.com/AnjulaMehto/MCA/blob/main/company_sales_data.csv 
##### Apply ‘seaborn’ library to do the following. 
###### 1. Read Total profit of all months and show it using a line plot. 
###### 2. Read all product sales data and show it using a multiline plot. 
###### 3. Plot bar chart for all the features/attributes.

```python
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
plt.figure(figsize=(8, 5))

sns.lineplot(x="month_number", y="total_profit", data=df, marker="o")

plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(x="month_number", y="facecream", data=df, label="Face Cream")
sns.lineplot(x="month_number", y="facewash", data=df, label="Face Wash")
sns.lineplot(x="month_number", y="toothpaste", data=df, label="Toothpaste")
sns.lineplot(x="month_number", y="bathingsoap", data=df, label="Bathing Soap")
sns.lineplot(x="month_number", y="shampoo", data=df, label="Shampoo")
sns.lineplot(x="month_number", y="moisturizer", data=df, label="Moisturizer")

plt.title("Product Sales per Month")
plt.xlabel("Month Number")
plt.ylabel("Sales")
plt.grid()

plt.show()
```

```python
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

total_sales = df[products].sum()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=products,
    y=total_sales.values,
    hue=products,
    legend=False
)

plt.title("Total Sales of All Products")
plt.xlabel("Products")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.grid(axis="y")

plt.show()
```
