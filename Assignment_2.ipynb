#### 3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

```python
import random
```

##### i.Generate a list of 100 random numbers between 100 and 900 (inclusive). 

```python
random.seed(1024170165)
numbers = [random.randint(100, 900) for _ in range(100)]
```

##### ii. Count and print all odd numbers in the list. 

```python
odd_numbers = [num for num in numbers if num % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Number of odd numbers:", len(odd_numbers))
```

##### iii. Count and print all even numbers in the list.

```python
even_numbers = [num for num in numbers if num % 2 == 0]
print("Even numbers:", even_numbers)
print("Number of even numbers:", len(even_numbers))
```

##### iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True

prime_numbers = [num for num in numbers if is_prime(num)]

print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))
```

##### v. Print the number that occurs most frequently in your list, and how many times it occurs. 

```python
frequency = {num: numbers.count(num) for num in numbers}

most_frequent_number = max(frequency, key=frequency.get)
frequency_count = frequency[most_frequent_number]

print("Most frequent number:", most_frequent_number)
print("It occurs", frequency_count, "times.")
```

#### 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
digits = [1,0,2,4,1,7,0,1,6]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
```

##### vi. Find the union of A and B — all unique values across both sets.

```python
union_set = A.union(B)
print("Union of A and B:", union_set)
```

##### vii. Find the intersection of A and B — values common to both.

```python
intersection_set = A.intersection(B)
print("Intersection of A and B:", intersection_set)
```

##### viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference().

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
```

##### ix. Find the symmetric difference of A and B.

```python
symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)
```

##### x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```
